## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
**What this work CAN say:**
- Observed: in this sample, certain features (position tier, content age,
  impressions) show a measurable relationship with which pages later decline.
- Directional: pages with these characteristics are *more likely* to be
  worth reviewing first, based on patterns in past data.
- Decision-support: a ranked queue that helps a content team prioritize their
  limited review time — a recommendation, not a verdict.

**What this work will NEVER say:**
- Not causal proof — a correlation between content age and decline doesn't
  mean age *causes* decline; some third factor could explain both.
- Not "predicting Google's algorithm" — the model reads patterns in
  observable signals, not the search engine's actual ranking logic.
- Not a guarantee for any single page — precision numbers describe average
  performance across a batch, not certainty for one page pulled from the list.
```
